## RMSE

**Что показывает:**

Среднюю величину ошибки, но с сильным штрафом за большие промахи.


**За что отвечает:**

насколько далеко прогноз от истины, особенно если большие ошибки очень нежелательны.


## MAE

**Что показывает:**

Среднюю абсолютную ошибку.

**За что отвечает:**

Более устойчива к выбросам, чем RMSE.

## Huber 

**Идея:**

- маленькие ошибки штрафует как квадрат,
- большие — как модуль.

**Что показывает:**

Компромисс между RMSE и MAE.

**За что отвечает:** 

выбросы важны, но ты не хочешь, чтобы они полностью доминировали метрику. -> хорошая основная метрика для returns.

## Sign accuracy

Sign Accuracy= $\frac{1}{n} \Sigma 1 [sign(y_i) = sign(\hat{y_i})]$

**Что показывает:**

Как часто модель правильно угадывает направление: вверх или вниз.

**Интерпретация**
- 0.50 — примерно случайно
- $>$ 0.50 — уже есть directional signal
- 0.55+ — обычно уже интересно
- 0.60+ — очень хорошо для noisy financial returns

Очень важна, если модель потом используется для risk monitoring.


## Negative-day sign accuracy

**Что показывает**

Как часто модель правильно понимает знак именно на отрицательных днях.

**Почему это полезно:**

Если тебе важнее downside-risk, чем все дни подряд, это очень содержательная метрика.

## Tail RMSE 10% и 5%

Берутся только те дни, где истинные returns входят в нижние 10% и  5%.

**Что показывает:**

Насколько модель ошибается именно на стрессовых / плохих днях. Для 5% очень близко к логике VaR / CoVaR.

`Меньше — лучше.`


## Tail MAE 10% и 5%

То же самое, но через абсолютную ошибку.

**Что показывает:**

Среднюю ошибку на плохих днях без сверхсильного наказания за extreme outliers.

**Когда полезно**

Если tail RMSE слишком чувствителен к нескольким экстремальным дням. Очень близко к логике VaR / CoVaR.

## Downside hit rate

**Смысл:**

смотрим на реальные плохие дни, проверяем, как часто модель тоже пометила их как “плохие”. (Это дни, которые попали в квантиль 10%)

Это похоже на Recall для tail events.

**Интерпретация:**

Если высокий downside hit rate: модель хорошо ловит stress days.

`Больше — лучше.`

## Downside precision

**Смысл:**

из тех дней, которые модель назвала плохими, сколько реально оказались плохими.

Precision для tail events.

**Интерпретация:**

Если высокий precision: модель не слишком часто бьёт ложную тревогу.

`Больше — лучше.`

## Downside F1

Комбинирует: downside hit rate  и downside precision


**Зачем:**

Это удобный единый показатель качества “ловли плохих дней”.

`Больше — лучше.`

## MASE

Mean Absolute Scaled Error. - Сравнивает MAE с наивным one-step forecast.

**Что показывает:**

Насколько  модель лучше или хуже наивного предсказания.


**Очень важная интерпретация:**
- MASE < 1 — модель лучше наивного прогноза
- MASE = 1 — примерно как наивный
- MASE > 1 — хуже наивного

**Почему полезно:** Это scale-free метрика, удобно сравнивать между периодами и тикерами.

`Меньше — лучше.`
