## CoVaR

**Волатильность компании и рынка:**

$$
\sigma_i(t) = \operatorname{Std}\big(r_i(t-w+1), \dots, r_i(t)\big)
$$

$$
\sigma_m(t) = \operatorname{Std}\big(r_m(t-w+1), \dots, r_m(t)\big)
$$

где:

$r_i(t)$ — доходность компании,

$r_m(t)$ — доходность рынка,

w — rolling window.

**Downside beta**

$$
\beta_i^{-}(t)= \frac{\operatorname{Cov}\big(r_i, r_m \mid r_m < 0\big)}{\operatorname{Var}\big(r_m \mid r_m < 0\big)}
$$

То есть мы оцениваем чувствительность компании к рынку только в плохие рыночные дни.

**Рыночный VaR proxy**

Для выбранного уровня q используем нормальный квантиль $z_q$, например для q=0.05: $z_q \approx −1.645$

$$
VaR_m^{proxy}(t) = z_q \cdot \sigma_m(t)
$$

**Company VaR proxy**

Если у нас есть прогноз доходности компании $\hat{r}_i(t)$, то:

$$
VaR_i^{proxy}(t) = \hat r_i(t) + z_q \cdot \sigma_i(t)
$$

А если хотим считать по истинным доходностям, то:

$$
VaR_{i,true}^{proxy}(t) = r_i(t) + z_q \cdot \sigma_i(t)
$$

## ΔCoVaR proxy

$$
\Delta CoVaR_i^{proxy}(t)
=
CoVaR_i^{proxy}(t) - VaR_i^{proxy}(t)
$$

## Чем это отличается от Adrian & Brunnermeier

У Adrian & Brunnermeier идея такая

Они определяют CoVaR как Value-at-Risk финансовой системы условно на том, что конкретный институт находится в distress.

То есть по смыслу:

$$
Pr\big(R^{system} \le CoVaR_q^{system \mid i}\; \big|\; R^i = VaR_q^i\big)=q
$$

А ΔCoVaR — это разница между:

CoVaR системы, когда институт i в distress, и CoVaR системы в его нормальном состоянии.

$$
\Delta CoVaR_q^i
=
CoVaR_q^{system \mid i \; in \; distress}
-
CoVaR_q^{system \mid i \; in \; median \; state}
$$

Цель диплома:

цель — прогнозировать returns компании и потом на их основе строить risk proxies.

У Adrian & Brunnermeier:

цель — измерить systemic spillover / contribution to systemic risk. То есть у них CoVaR — это “насколько distress этой фирмы сдвигает вниз хвост доходности системы”.

В дипломе текущая конструкция скорее отвечает на вопрос: “как выглядит downside risk компании с учётом рыночного стресса”.


### Интервалы на графиках

1. Нижний 5% квантиль

Формально это значение $q_{0.05}$, такое что:

$$
P(Y \le q_{0.05}) = 0.05
$$

**Смысл:** только в 5% случаев реальное значение будет хуже этой границы. Для доходностей это очень похоже на one-sided downside bound и по смыслу близко к VaR.

2. 90% prediction interval

Это уже двусторонний интервал, например между 5% и 95% квантилями:

$$
[q_{0.05}, q_{0.95}]
$$

**Смысл:** модель считает, что с вероятностью около 90% будущее значение попадёт внутрь этого диапазона.

Для risk-задачи обычно полезнее именно:

нижний 5% квантиль,
нижний 1% квантиль,
или нижняя граница 90% интервала.

Потому что тебя интересует именно левый хвост, а не симметричная uncertainty band.
